```python
import time # importing time module
from calendar import isleap # importing for leap year calculation
```

```python
# def a function to check for a leap year
def judge_leap_year(year):
    if isleap(year):
        return True
    else:
        return False
```

```python
# logic to cal the days in months
def month_days(month, leap_year):
    if month in [1, 3, 5, 7, 8, 10, 12]:
        return 31
    elif month in [4, 6, 9, 11]:
        return 30
    elif month == 2 and leap_year:
        return 29
    elif month == 2 and (not leap_year):
        return 28
```

```python
name = input("Please enter your name:  ") # input your name here
age = input("Please enter your age:  ")  # input your age
localtime = time.localtime(time.time()) #this line gets the current date and time from the system.

year = int(age) # here the age in year is already discovered
month = year*12 + localtime.tm_mon # this line calculates the age in months
day = 0 # here the day variable is assigned 0 value so it can be use for calculation.

begin_year = int(localtime.tm_year) - year # this line finds the begin year of calculation i.e. one year after birth.
end_year = begin_year + year # this line gives us the same result as int(localtime.tm_year)

for y in range(begin_year,end_year):
    if (judge_leap_year(y)):
        day = day + 366
    else:
        day = day + 365


leap_year = judge_leap_year(localtime.tm_year)
for m in range(1, localtime.tm_mon):
    day = day + month_days(m, leap_year)


day = day + localtime.tm_mday
print(f"\n\t{name}'s age is {year} years or ", end="")
print(f"{month} months or {day} days")
```
